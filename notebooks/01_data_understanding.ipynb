<h4>Outlier Detection (Boxplots)</h4>

```python
# Identify numerical features for boxplot analysis
numerical_features = ['BMI', 'MentHlth', 'PhysHlth', 'Age']

# Create boxplots for numerical features
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()

for idx, col in enumerate(numerical_features):
    axes[idx].boxplot(df[col].dropna())
    axes[idx].set_title(f'Boxplot of {col}')
    axes[idx].set_ylabel(col)
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Statistical summary to identify outliers
for col in numerical_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers ({len(outliers)/len(df)*100:.2f}%)")
```

**Insights:**
- Outliers are most frequent in **PhysHlth (16.14%)** and **MentHlth (14.27%)**, less in **BMI (3.88%)**, none in **Age** (IQR method).

<h4>Bivariate Analysis</h4>

```python
def plot_cont_feature_by_diabetes(
    df: pd.DataFrame, 
    features: List[str], 
    diabetes_col: str = 'Diabetes_012'
) -> None:
    """
    Plot feature distributions grouped by diabetes status.
    
    Args:
        df: DataFrame containing the data
        features: List of feature column names to plot
        diabetes_col: Name of diabetes target column (default: 'Diabetes_012')
    """
    fig, axes = plt.subplots(1, len(features), figsize=(5*len(features), 5))
    axes = axes if len(features) > 1 else [axes]
    
    diabetes_values: List[int] = sorted(df[diabetes_col].unique())
    diabetes_labels: List[str] = ['No Diabetes', 'Prediabetes', 'Diabetes']
    
    for ax, col in zip(axes, features):
        data: List = [df[df[diabetes_col] == val][col].dropna() for val in diabetes_values]
        ax.boxplot(data, labels=diabetes_labels)
        ax.set_title(f'{col} by Diabetes Status')
        ax.set_ylabel(col)
        ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
plot_cont_feature_by_diabetes(df,['BMI'])
```

```python
plot_cont_feature_by_diabetes(df,['PhysHlth'])
```

```python
plot_cont_feature_by_diabetes(df,['Age'])
```

```python
def plot_binary_features_by_diabetes(
    df: pd.DataFrame,
    features: List[str],
    diabetes_col: str = "Diabetes_012",
    normalize: bool = False
) -> None:
    """
    Plot grouped bar charts for binary features against diabetes status.

    Args:
        df: Input DataFrame.
        features: List of binary feature column names (values 0/1).
        diabetes_col: Target column name.
        normalize: If True, plot proportions instead of counts.
    """
    n_cols = 2
    n_rows = (len(features) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 4 * n_rows))
    axes = axes.ravel() if len(features) > 1 else [axes]

    for ax, feature in zip(axes, features):
        ct = pd.crosstab(df[feature], df[diabetes_col], normalize="index" if normalize else False)
        ct = ct.sort_index()
        ct.plot(kind="bar", ax=ax, color=["green", "orange", "red"], alpha=0.8)
        ax.set_title(f"{feature} vs {diabetes_col}")
        ax.set_xlabel(f"{feature} (0=No, 1=Yes)")
        ax.set_ylabel("Proportion" if normalize else "Count")
        ax.legend(["No Diabetes", "Prediabetes", "Diabetes"], title="Diabetes Status")
        ax.grid(True, axis="y", alpha=0.3)
        ax.set_xticklabels(["0", "1"], rotation=0)

    # Hide any unused subplots
    for ax in axes[len(features):]:
        ax.set_visible(False)

    plt.tight_layout()
    plt.show()

# Example usage:
# plot_binary_features_by_diabetes(df, ["HighBP", "HighChol", "Smoker", "PhysActivity"], normalize=False)
```

```python
plot_binary_features_by_diabetes(df, ["HighBP", "HighChol", "Smoker", "PhysActivity"], normalize=True)
```

**Insights:**
- Mean values increase with diabetes status:
  - **BMI:** 27.74 → 30.72 → 31.94  
  - **Age:** 7.79 → 9.08 → 9.38  
  - **PhysHlth:** 3.58 → 6.35 → 7.95  
- Diabetes prevalence is higher among **HighBP** and **HighChol** groups (e.g., **Diabetes = 24.45% with HighBP vs 6.04% without**).
